```python
import os

# Directories
repo_dir = "/content/Marigold"
input_dir = os.path.join(repo_dir, "input")
output_dir = os.path.join(repo_dir, "output")
output_dir_color = os.path.join(output_dir, "depth_colored")
output_dir_tif = os.path.join(output_dir, "depth_bw")
output_dir_npy = os.path.join(output_dir, "depth_npy")

os.chdir(repo_dir)
```

```python
from marigold import MarigoldPipeline

pipe = MarigoldPipeline.from_pretrained(ckpt_path)

pipe = pipe.to("cuda")
```

```python
denoising_steps = 4
ensemble_size = 5
processing_res = 768
match_input_res = True

import numpy as np
import torch
from tqdm.auto import tqdm
from PIL import Image

image_path = "/content/12 (1).png"
input_image = Image.open(image_path)

with torch.no_grad():
    pipeline_output = pipe(
        input_image,
        denoising_steps=denoising_steps,
        ensemble_size=ensemble_size,
        processing_res=processing_res,
        match_input_res=match_input_res,
        batch_size=0,
        color_map= "Spectral",
        show_progress_bar=True,
    )

    depth_pred = pipeline_output.depth_np
    depth_colored = pipeline_output.depth_colored
```

```python
np.save("/content/depth_map.npy", depth_pred)
depth_colored.save("/content/depth_map_colored.png")
```

```python
depth_to_save = (depth_pred * 65535.0).astype(np.uint16)
```

```python
depth_to_save.min(),depth_to_save.max()
```

```python
true_depth = Image.open("/content/12.png")
depth_arr = np.array(true_depth)
```

```python
depth_arr.min(),depth_arr.max()
```

```python
depth_to_save = depth_to_save.astype(np.float32)

min_GT, max_GT = depth_arr.min(), depth_arr.max()
min_pred, max_pred = depth_to_save.min(), depth_to_save.max()

depth_to_save_rescaled = (depth_to_save - min_pred) * ((max_GT - min_GT) / (max_pred - min_pred)) + min_GT
depth_to_save_rescaled = np.round(depth_to_save_rescaled).astype(np.uint16)
print(depth_to_save_rescaled.min(), depth_to_save_rescaled.max())  # Should match depth_arr range
```

```python
# Convert to PIL Image and save as black and white PNG
depth_image = Image.fromarray(depth_to_save_rescaled)
depth_image.save("/content/depth_map_bw.png")  # Save as grayscale PNG
```

```python
print(depth_to_save.shape, depth_to_save.dtype)
print(depth_arr.shape, depth_arr.dtype)
```

```python
depth_to_save.min()
```

```python
# Convert depth maps to float to avoid division issues
pred = depth_to_save_rescaled.astype(np.float32)
gt = depth_arr.astype(np.float32)

# Absolute Relative Error (Abs Rel)
abs_rel = np.mean(np.abs(pred - gt) / gt)

# Squared Relative Error (Sq Rel)
sq_rel = np.mean((pred - gt) ** 2 / gt)/(np.mean(gt))

# Root Mean Squared Error (RMSE)
rmse = np.sqrt(np.mean((pred - gt) ** 2))
nrmse = rmse/(np.mean(gt))

# RMSE log
rmse_log = np.sqrt(np.mean((np.log(pred) - np.log(gt)) ** 2))

# Print results
print(f"Abs Rel: {abs_rel:.4f}")
print(f"RMSE: {nrmse:.4f}")
print(f"RMSE log: {rmse_log:.4f}")
print(f"Sq Rel: {sq_rel:.4f}")
```
